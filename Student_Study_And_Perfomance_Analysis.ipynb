```python
#STUDENT STUDY AND PERFOMANCE ANALYSIS
import matplotlib.pyplot as plt

students = ["Ann", "Brian", "Carol", "David", "Eric",
            "Faith", "George", "Hellen", "Ian", "Jane"]

study_hours = [2, 3, 4, 5, 6, 7, 8, 4, 6, 9]

scores = [50, 55, 62, 68, 72, 78, 85, 60, 75, 92]

# LINE PLOT
fig, ax=plt.subplots(figsize=(12,6))

ax.plot(students,scores,
        marker='o')

ax.set_title('Student Scores')
ax.set_xlabel('Student')
ax.set_ylabel('Scores')
ax.tick_params(axis='x',rotation=45)

# Finding student with the highest score
highest_score=max(scores)
highest_index=scores.index(highest_score)
highest_student=students[highest_index]
ax.annotate(f'Highest Score: {highest_student} ({highest_score})',
    xy=(highest_student,highest_score),
    xytext=('Ian',75),
        arrowprops=dict(arrowstyle='->')
    )
# Finding the student with the lowest score
lowest_score=min(scores)
lowest_index=scores.index(lowest_score)
lowest_student=students[lowest_index]
ax.annotate(f'Lowest Score: {lowest_student} ({lowest_score})',
            xy=(lowest_student,lowest_score),
            xytext=('Brian',55),
            arrowprops=dict(arrowstyle='->'))
ax.grid()
plt.tight_layout()
plt.show()
```

```python
#PART B- SCATTER PLOT
import matplotlib.pyplot as plt
import numpy as np
students = ["Ann", "Brian", "Carol", "David", "Eric",
            "Faith", "George", "Hellen", "Ian", "Jane"]

study_hours = [2, 3, 4, 5, 6, 7, 8, 4, 6, 9]

scores = [50, 55, 62, 68, 72, 78, 85, 60, 75, 92]

# Scatter plot showing study hours  vs score
fig, ax=plt.subplots(figsize=(12,6))

ax.scatter(study_hours,
           scores,
           s=100,
           alpha=0.7)

ax.set_title('Study Hours vs Scores')
ax.set_xlabel('Study Hours')
ax.set_ylabel('Scores')
 # Calculating correlation
correlation=np.corrcoef(study_hours,scores)[0,1]
print(f'Correlation: ',correlation)
ax.grid()
plt.show()
```
